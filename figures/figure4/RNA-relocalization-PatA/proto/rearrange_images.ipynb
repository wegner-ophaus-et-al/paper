```python
from pathlib import Path
import shutil
```

```python
roots = [Path("/Users/julian/local_files/pata_kim_data/RNAscope_tdrd7_nanos/2022-11-18_antibody-Vasa_RNAscope-tdrd7_nanos_Dnd-GFP_10hpf_PatA"),
         Path("/Users/julian/local_files/pata_kim_data/RNAscope_tdrd7_nanos/2022-12-16_antibody-Vasa_RNAscope-tdrd7_nanos_10hpf_PatA")]
out_dir = Path("/Users/julian/local_files/pata_kim_data/RNAscope_tdrd7_nanos/collected_data")
```

```python
for root in roots:
    list_dirs = sorted([d for d in root.iterdir() if d.is_dir()])
    # Go over condition dirs and copy all lsm files to the collection dir
    for working_dir in list_dirs:
        date = root.name.split("_")[0]
        condition = working_dir.name
        image_paths = sorted(working_dir.glob("*.lsm"))
        for img_path in image_paths:
            new_name = f"{date}_{condition}_{img_path.name}"
            shutil.copy(img_path, out_dir / new_name)
```
